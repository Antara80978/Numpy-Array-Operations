# **NUMPY ARRAY OPERATIONS**

```python
import numpy as np
```

```python
# Create any two arrays

arr1 = np.array([1 , 2 , 3 , 4 , 5 , 6 , 7 ])
arr2 = np.array([8 , 9 , 10 , 11 , 12 , 13 , 14])
```

```python
# Display the arrays

print("First Array = " ,arr1);
print("Second Array = ", arr2)
```

```python
#Check the Shape of both the arrays

print("Shape of First Array = "  , arr1.shape)
print("Shape of Second Array = " ,  arr2.shape)
```

```python
#Addition  of arrays

Addition = np.array(arr1 + arr2)
print ("Addition of two arrays = " , Addition)
```

```python
# Subtraction of array

Subtraction = np.array( arr1 - arr2)
print (" Subtraction of two arrays =" , Subtraction)
```

```python
#Product of arrays
multiplication = arr1 * arr2
print ( "Product of given arrays = " , multiplication)
```

```python
# Division of array

Division = arr1 / arr2
print ("Division of given array =", Division)
```
